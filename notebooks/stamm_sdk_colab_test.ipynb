```python
from stamm_sdk import Client
from getpass import getpass

BASE_URL = "http://188.245.241.118:8080"

email = input("Email: ")
password = getpass("Password (hidden): ")

sdk = Client(base_url=BASE_URL)
login_response = sdk.login(email=email, password=password)
print("Logged in. Token keys:", list(login_response.keys()))
```

```python
# List projects and pick one
projects = sdk.ml.list_projects()
projects
```

```python
# Set these values based on your registry
project_id = input("Project ID: ")
models = sdk.ml.list_models(project_id)
models
```

```python
# Prediction test
model_id = input("Model ID: ")
features = {
    "temperature": 24.7,
    "ph": 6.8
}
prediction = sdk.ml.predict(project_id=project_id, model_id=model_id, features=features)
prediction
```

```python
# Cleanup
sdk.logout()
sdk.close()
print("Done")
```
